```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

preprocessor = ColumnTransformer(
        transformers = [
            ("num", StandardScaler(), numerical_features),
            ("cat", OneHotEncoder(handle_unknown = "ignore"), categorical_features)
        ]
)
```

## 10. Threshold Tuning

```python
import numpy as np
```

```python
thresholds = [0.3, 0.4, 0.5, 0.6]

for threshold in thresholds:
    y_pred_threshold = np.where(
        y_prob_logistic_tuned >= threshold,
        "Yes",
        "No"
    )

    print(f"\nThreshold: {threshold}")
    print("Precision:", precision_score(y_test, y_pred_threshold, pos_label="Yes"))
    print("Recall:", recall_score(y_test, y_pred_threshold, pos_label="Yes"))
    print("F1:", f1_score(y_test, y_pred_threshold, pos_label="Yes"))
```

```python
from sklearn.model_selection import cross_val_predict

y_train_prob_cv = cross_val_predict(
    logistic_grid_search.best_estimator_,
    X_train,
    y_train,
    cv=cv,
    method="predict_proba",
    n_jobs=-1
)[:, 1]
```

```python
threshold_results = []

for threshold in np.arange(0.10, 0.91, 0.01):
    y_pred_threshold = np.where(
        y_train_prob_cv >= threshold,
        "Yes",
        "No"
    )

    precision = precision_score(
        y_train,
        y_pred_threshold,
        pos_label="Yes"
    )

    recall = recall_score(
        y_train,
        y_pred_threshold,
        pos_label="Yes"
    )

    f1 = f1_score(
        y_train,
        y_pred_threshold,
        pos_label="Yes"
    )

    threshold_results.append(
        (threshold, precision, recall, f1)
    )
```

```python
best_threshold_result = max(
    threshold_results,
    key=lambda x: x[3]
)

print("Best threshold:", best_threshold_result[0])
print("Precision:", best_threshold_result[1])
print("Recall:", best_threshold_result[2])
print("F1:", best_threshold_result[3])
```

```python
best_threshold = best_threshold_result[0]

y_pred_logistic_threshold = np.where(
    y_prob_logistic_tuned >= best_threshold,
    "Yes",
    "No"
)
```

```python
print("Threshold:", best_threshold)
print("Accuracy:", accuracy_score(y_test, y_pred_logistic_threshold))
print("Precision:", precision_score(
    y_test, y_pred_logistic_threshold, pos_label="Yes"
))
print("Recall:", recall_score(
    y_test, y_pred_logistic_threshold, pos_label="Yes"
))
print("F1:", f1_score(
    y_test, y_pred_logistic_threshold, pos_label="Yes"
))
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_logistic_threshold))
```

```python
new_customer = pd.DataFrame([{
    "gender": "Female",
    "SeniorCitizen": 0,
    "Partner": "Yes",
    "Dependents": "No",
    "tenure": 12,
    "PhoneService": "Yes",
    "MultipleLines": "No",
    "InternetService": "Fiber optic",
    "OnlineSecurity": "No",
    "OnlineBackup": "Yes",
    "DeviceProtection": "No",
    "TechSupport": "No",
    "StreamingTV": "Yes",
    "StreamingMovies": "Yes",
    "Contract": "Month-to-month",
    "PaperlessBilling": "Yes",
    "PaymentMethod": "Electronic check",
    "MonthlyCharges": 85.0,
    "TotalCharges": 1020.0
}])

new_customer
```

```python
churn_probability = loaded_model.predict_proba(new_customer)[0, 1]

print("Churn probability:", churn_probability)
```

```python
prediction = np.where(
    churn_probability >= loaded_threshold,
    "Yes",
    "No"
)

print("Predicted churn:", prediction)
```

```python
def predict_churn(customer_data):
    probability = loaded_model.predict_proba(customer_data)[0, 1]

    prediction = (
        "Yes" if probability >= loaded_threshold
        else "No"
    )

    return prediction, probability
```
